```python
domingos = datos[datos["fecha_hora"].dt.dayofweek == 0]
print(domingos["demanda_kW"].mean())
```

```python
domingos = datos[datos["fecha_hora"].dt.dayofweek <= 4]
print(domingos["demanda_kW"].mean())
```

```python
enero = datos[datos["fecha_hora"].dt.month == 1]
print(enero["demanda_kW"].mean())
```

```python
enero = datos[datos["fecha_hora"].dt.month == 1]
print(enero[enero["fecha_hora"].dt.hour == 15]["demanda_kW"].mean())
```

```python
Consumo_mensual = datos.groupby(datos["fecha_hora"].dt.month)["demanda_kW"].sum()/1000
print(Consumo_mensual.idxmax())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(datos["demanda_kW"])
plt.show()
```

```python
plt.xlabel("Hora del día")
plt.ylabel("Demanda (kW)")
plt.title("Perfil de carga - día laboral típico")
```

```python
plt.bar(Consumo_mensual.index, Consumo_mensual.values)
plt.xticks(Consumo_mensual.index, ["Ene","Feb","Mar","Abr","May","Jun","Jul","Ago","Sep","Oct","Nov","Dic"])
plt.grid(axis="y", alpha=0.3)
plt.title("Consumo mensual de energía - planta industrial 2025")
plt.ylabel("Energía (MWh)")
plt.xlabel("Mes")
plt.show()
```
